```python
class Shape:
  def area(self):
    pass
  def perimeter(self):
    pass

class circle(Shape):
  def __init__(self,r):
    self.r=r
  def area(self):
    return self.r*self.r*3.14
class traingle(Shape):
  def __init__(self,a,b,c,h):
    self.b=b
    self.h=h
    self.a=a
    self.c=c
  def area(self):
    return 0.5*self.b*self.h
  def perimeter(self):
    return self.a+self.b+self.c
class square(Shape):
  def __init__(self,s):
    self.s=s
  def area(self):
    return self.s*self.s
  def perimeter(self):
    return 4*self.s
s1=circle(2)
print(s1.area())
```
