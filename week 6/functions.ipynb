##Checking palindrome or not

```python
def p(a):
  n=a
  rev=0
  while(a>0):
    r=int(a%10)
    rev=rev*10+r
    a=a//10
  if(n==rev):
    return True
  else:
    return False
x = int(input())
p(x)
print(p(2546))
```

##reverse a given string using functions

```python
def s(a):
  n=a
  b=len(a)
  rev=""
  for i in range(b-1,-1,-1):
     rev= rev+a[i]
  if(n==rev):
    return True
  else:
    return False
x=input()
s(x)
```

##example on default arguments

```python
def msg(name,branch="ai & ds"):
  print("name is ",name,"branch is",branch)
name=input()
branch=input()
msg(name)
```

##example on keyword arguments

```python
def msg(name,branch="ai & ds"):
  print("branch=",branch,"name=",name)
name=input()
msg(name)
```

## example on variable length

```python
def add(*num):
  print(num)
  print(sum(num))
add(1,2,3,4)
```

##function that returns multiple values

```python
def m(*num):
  s=sum(num)
  c=len(num)
  a=s/c
  return s,c,a
res_sum,res_count,res_avg=m(3,5,6)
print(f"sum is {res_sum},count is {res_count},average is {res_avg}")
```
