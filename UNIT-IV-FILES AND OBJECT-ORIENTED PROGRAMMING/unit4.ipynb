```python
class Bank:
    def __init__(self, b):
        self.b = b
    def add(self, v):
        self.b += v
acc = Bank(5000)
acc.add(1000)
print("3. Balance:", acc.b)
```

```python
class Circle:
    def __init__(self, r):
        self.r = r
    def area(self):
        return 3.14159 * self.r ** 2
c = Circle(5)
print("4. Circle Area:", c.area())
```

```python
class Student:
    def __init__(self, n, r):
        self.n, self.r = n, r
s = Student("Jagadeesh", "25PA1A4598")
print(f"6. Student -> Name: {s.n}, Roll: {s.r}")
```
