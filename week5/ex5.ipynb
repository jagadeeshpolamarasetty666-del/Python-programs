##counting no.of vowels and consonants

```python
a=input()
b=a.lower()
c=0
for i in range(len(b)):
  if(b[i]=='a' or b[i]=='e' or b[i]=='i' or b[i]=='o' or b[i]=='u'):
    c=c+1
print("vowels=",c)
print("consonants=",len(b)-c)
```
