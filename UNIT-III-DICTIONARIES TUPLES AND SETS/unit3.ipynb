```python
tuple1 = ("Jagadeesh", 21, "Bhimavaram")
tuple2 = ("Vishnu Institute of Technology", "Rahul", 22)
print("1. Concatenated Tuple:", tuple1 + tuple2)
```

```python
text = "Vishnu Institute of Technology"
print("2. Vowel Count:", sum(text.count(v) for v in "aeiouAEIOU"))
```

```python
student = {"name": "Jagadeesh", "roll": "25PA1A4598"}
student["cgpa"] = 9.2
marks = {"Maths": 95, "Python": 98, "DSA": 90}
print("3. Dict Ops:", "roll" in student, student, sum(marks.values()))
```

```python
a, b = {1, 2, 3, 4}, {3, 4, 5, 6}
print("4. Set Ops:", a | b, a & b)
```

```python
freq = {c: "Python".count(c) for c in "Python"}
print("5. Freq:", freq)
```

```python
scores = {"Python": 98, "Maths": 92, "Chemistry": 88}
print("6. Sorted:", sorted(scores.items(), key=lambda x: x[1]))
```
