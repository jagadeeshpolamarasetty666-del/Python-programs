```python
tuple1 = ("Jagadeesh", 21, "Bhimavaram")
tuple2 = ("Vishnu Institute of Technology", "Rahul", 22)
concatenated_tuple = tuple1 + tuple2
print("1. Concatenated Tuple & Unpacking:", concatenated_tuple)
name, age, city = tuple1
print(f"Name: {name}, Age: {age}")
```

```python
text = "Vishnu Institute of Technology"
vowels = "aeiouAEIOU"
vowel_count = sum(text.count(v) for v in vowels)
print("2. Vowel Count in String:", vowel_count)
```

```python
student = {"name": "Jagadeesh", "roll": "25PA1A4598", "branch": "CSE-AI&DS"}
print("3. Key \"roll\" exists:", "roll" in student)
student["cgpa"] = 9.2
print("Updated Dictionary:", student)
marks = {"Maths": 95, "Python": 98, "DSA": 90}
print("Sum of Marks:", sum(marks.values()))
```

```python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
print("4. Set Union:", set_a | set_b)
print("Set Intersection:", set_a & set_b)
print("Set Difference:", set_a - set_b)
```

```python
word = "Python"
freq = {}
for char in word:
    freq[char] = freq.get(char, 0) + 1
print("5. Character Frequency Counter:", freq)
```

```python
scores = {"Python": 98, "Maths": 92, "Chemistry": 88}
sorted_scores = sorted(scores.items(), key=lambda item: item[1])
print("6. Sorted Dictionary by Value:\n", sorted_scores)
```
