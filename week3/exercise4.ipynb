#functions:
sum,max,len,min

```python
L=[4,"hayato",160.5,45,25,132,34]
print(L[2]) #indexing
#slicing
print(L[0:])
print(L[:7])
print(L[2:6])
print(L[7:0:-1])
print(L[2:7:2])
print(L[7:0:-2])
```

```python
L1=[2,4,5,3,7,2,8]
L2=[56,87,46,743,83,28]
# Predefined Functions
print(sum(L1))
print(max(L1))
print(L1+L2) #concatenation
print(min(L2))
print(len(L1)+len(L2))
```

```python
#membership operator
2 in L1
```

```python
#iteration
for i in L1:
  print(i,end=" ")
```
